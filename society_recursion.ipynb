Write a recursive total_units(block) using isinstance(item, list). It must return 1043 for the whole society and 353, 425 and 265 for the wings. 2. <p>Write a recursive nesting_depth(block, level=1) that returns the deepest nesting level of society_units (it must return 4, and 1 for [120, 135, 98]). Add a print at the start of every call that indents 4 spaces per level and shows the block, so the call stack is visible in the output. Run it on the whole society and keep the trace visible in the notebook.</p>
3. <p>Finish with a scope demo: set a global RATE_PER_UNIT = 7 and write bill_for_wing(block) that uses a local wing_bill and increments a global counter bills_generated with the global keyword. Call it for the three wings; bills must be 2471, 2975 and 1855 (total 7301) and the counter must end at 3. In the last cell, print wing_bill outside the function so the NameError stays visible.
```python
society_units = [[120, 135, 98],[210, [75, 80], 60],[150, [[40, 45], 30]]]
print(society_units)
```

```python
def total_units(block):
  total = 0
  for item in block:
    if isinstance(item, list):
      total += total_units(item)
    else: 
      total += item
  return total

print("Whole Society:", total_units(society_units))
for wing_block, wing in enumerate(society_units, start=1):
  print(f"Wing {wing_block}:", total_units(wing))
```

```python
def nesting_depth(block, level=1):
    print(" " * (4 * (level - 1)) + f"level {level}: {block}")

    if not isinstance(block, list):
        return level

    if not block:
        return level

    return max(
        nesting_depth(item, level + 1) if isinstance(item, list) else level
        for item in block
    )


print("Trace for the whole society:")
society_depth = nesting_depth(society_units)
print("Deepest nesting level:", society_depth)

print("\nDepth of a flat list:")
print(nesting_depth([120, 135, 98]))
```

```python
RATE_PER_UNIT = 7
bills_generated = 0


def bill_for_wing(block):
    global bills_generated
    wing_bill = total_units(block) * RATE_PER_UNIT
    bills_generated += 1
    return wing_bill


wing_bills = []
for wing_number, wing in enumerate(society_units, start=1):
    bill = bill_for_wing(wing)
    wing_bills.append(bill)
    print(f"Wing {wing_number} bill: {bill}")

print("Total bill:", sum(wing_bills))
print("Bills generated:", bills_generated)
```
