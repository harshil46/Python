```python
age = 23
height = 170.5
name = "Harshil"
spotify = True

print("Age:" ,age)
print(type(age))
print("--------------")
print("Height: " ,height)
print(type(height))
print("--------------")
print("Name: " ,name)
print(type(name))
print("--------------")
print("You have Spotify Account? - " ,spotify)
print(type(spotify))
print("--------------")
```

```python
def total_cart_amount(prices):
  total = 0.0

  for price in prices:
    total += float(price)

  return total

cart_prices = ["450.25","50", "65.50"]

print("Total Cart Amount: ", total_cart_amount(cart_prices))
```

```python
cric_score = input("Enter the Score: ")
print(cric_score)
print(type(cric_score))
print("--------------")
cric_score = int(cric_score)
print(cric_score)
print(type(cric_score))
print("--------------")

if cric_score >= 50:
  print("Half-Century!")
else:
  print("Keep Going!")
```
