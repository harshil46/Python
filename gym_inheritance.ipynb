1. <p>Create 'gym_inheritance.ipynb' with a base class Member (name, age) whose monthly_fee() returns 1200 and describe() returns a text line. Add subclasses StudentMember and PremiumMember using single inheritance and super().__init__(); PremiumMember also takes trainer_name. Create one object of each and print describe() for all three to show the inherited method works without rewriting it.</p>
2. <p>Override monthly_fee() so StudentMember pays 20% less (960) and PremiumMember pays 800 more (2000). Put one Member, one StudentMember and one PremiumMember in a list and call monthly_fee() on each in a loop, printing type(m).__name__ with each fee. Hint: Call super().monthly_fee() inside PremiumMember so the base fee is reused.</p>
3. <p>Finish with total_monthly_income(members) that sums monthly_fee() over any list of members. Build 8 members (3 Member, 3 StudentMember, 2 PremiumMember); it must return 10480. Use isinstance() to count the students (3) and print a final table of class name, count and income per class. Run Restart & Run All and submit.</p>

```python
class Member:
  def __init__(self, name, age):
    self.name = name
    self.age = age

  def monthly_fee(self):
    return 1200

  def describe(self):
    return f"Name: {self.name}, Age: {self.age}, Price: {self.monthly_fee()}"

class StudentMember(Member):
  def __init__(self, name, age):
    super().__init__(name, age)

  def monthly_fee(self):
    return super().monthly_fee() * 0.8

class PremiumMember(Member):
  def __init__(self, name, age, trainer_name):
    super().__init__(name, age)
    self.trainer_name = trainer_name

  def monthly_fee(self):
    return super().monthly_fee() + 800

m1 = Member("Aman", 22)
#print(m1.describe())
m2 = Member("Ruby", 26)
m3 = Member("Shriya", 24)

m4 = StudentMember("Rahul", 23)
#print(m2.describe())
m5 = StudentMember("Rohan", 25)
m6 = StudentMember("Vishwa", 24)

m7 = PremiumMember("Rajesh", 28, "Marsh")
#print(m3.describe())
m8 = PremiumMember("Manish", 27, "Vin")



members = [m1, m2, m3, m4, m5, m6, m7, m8]

# for m in members:
#   print(f"{type(m).__name__} : {m.monthly_fee()}")

def total_monthly_income(members):
  total = 0
  for m in members:
    total = total + m.monthly_fee()
  return total

student_count = 0

for m in members:
    if isinstance(m, StudentMember):
        student_count += 1

print("Number of students:", student_count)
print("Total monthly income:", total_monthly_income(members))

classes = ["Member", "StudentMember", "PremiumMember"]

print("\nClass Name Count Income")
print("-" * 45)

for class_name in classes:
    count = 0
    income = 0

    for m in members:
        if type(m).__name__ == class_name:
            count += 1
            income += m.monthly_fee()

    print(f"{class_name}, {count}, {income}")
```
