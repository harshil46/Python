```python
playlist_ids = [1, 5, 11, 21, 50]
print("Original: " ,playlist_ids)
print("------------------------------------------------")
playlist_ids.append(35)
print("Updated: " ,playlist_ids)
print("------------------------------------------------")
playlist_ids.extend([10, 25, 75])
print("Updated: " ,playlist_ids)
print("------------------------------------------------")
removed = playlist_ids.pop()
print("Removed: " ,removed)
print("Remain Playlist: " ,playlist_ids)
```

```python
insta_filters = ("first","second","third","forth")
print(insta_filters)
print("------------------------------------------------")
insta_filters = list(insta_filters)
insta_filters.remove("first")
print(insta_filters)
print("------------------------------------------------")
insta_filters.insert(0, "fifth")
print(insta_filters)
print("------------------------------------------------")
insta_filters = tuple(insta_filters)
print(insta_filters)
print("------------------------------------------------")
```
