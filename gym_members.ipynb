1. <p>Create 'gym_members.ipynb' and define a class Member with __init__(self, name, age, plan) storing name, age and plan ('Monthly', 'Quarterly' or 'Annual') and an empty list visits. Create five Member objects for a fictional gym you name, and print each object's name, age and plan using attribute access.</p>
2. <p>Add methods check_in(date_text) that appends to visits, visit_count() that returns len(visits), and plan_price() that returns 1200 for Monthly, 3300 for Quarterly and 11000 for Annual. Check in the first object on '2026-03-01', '2026-03-03' and '2026-03-05' and the second on two dates, then print visit_count() and plan_price() for all five objects. Hint: Create visits inside __init__, not at class level, so each object gets its own list.</p>
3. <p>Finish with six Member objects whose plans are Monthly, Monthly, Quarterly, Annual, Quarterly, Monthly. Add check-ins so the visit counts differ, then print total revenue (it must be 21200) and the name of the member with the most visits, and write the same lines to 'gym_report.txt'. Run Restart & Run All and submit. Constraint: Do not use inheritance yet.</p>

```python
class Member:
  def __init__(self, name, age, plan):
    self.name = name
    self.age = age
    self.plan = plan
    self.visit = []

  def check_in(self, date_text):
    self.visit.append(date_text)

  def visit_count(self):
    return len(self.visit)

  def plan_price(self):
    prices = {
      "Monthly" : 1200,
      "Quarterly" : 3300,
      "Yearly" : 11000
    }
    return prices[self.plan]

m1 = Member("Ayush", 21, "Monthly")
m2 = Member("Rajesh", 23, "Yearly")
m3 = Member("Ravi", 20, "Quarterly")
m4 = Member("Rahul", 25, "Yearly")
m5 = Member("Karan", 28, "Quarterly")

m1.check_in("2026-23-01")
m1.check_in("2026-23-03")
m1.check_in("2026-23-05")

m2.check_in("2026-23-01")
m2.check_in("2026-23-03")

members = [m1, m2, m3, m4, m5]

for member in members:
  # print("Name : ", member.name)
  # print("Age : ", member.age)
  # print("Plan : ", member.plan)
  # print("--------------------")
  print(f"{member.name}: Visits = {member.visit_count()}, Plan Price = {member.plan_price()}")
```

```python
class Member_two:
  def __init__(self, name, age, plan):
    self.name = name
    self.age = age
    self.plan = plan
    self.visit = []

  def check_in(self, date_text):
    self.visit.append(date_text)

  def visit_count(self):
    return len(self.visit)

  def plan_price(self):
      prices = {
        "Monthly" : 1200,
        "Quarterly" : 3300,
        "Annual" : 11000
      }
      return prices[self.plan]

m1 = Member_two("Aastha", 25, "Monthly")
m2 = Member_two("Rahil", 26, "Monthly")
m3 = Member_two("Karan", 22, "Quarterly")
m4 = Member_two("Pooja", 25, "Annual")
m5 = Member_two("Ishan", 27, "Quarterly")
m6 = Member_two("Aarzoo", 23, "Monthly")

m1.check_in("2026-03-01")
m1.check_in("2026-03-03")
m1.check_in("2026-03-05")

m2.check_in("2026-04-01")

m3.check_in("2026-05-05")
m3.check_in("2026-05-10")
m3.check_in("2026-05-15")
m3.check_in("2026-05-25")

m5.check_in("2026-04-10")
m5.check_in("2026-04-30")            
                     
members = [m1, m2, m3, m4, m5, m6]

report_lines = []

for member in members:
  line = f"{member.name}: Visits {member.visit_count()}, Price {member.plan_price()}"
  print(line)
  report_lines.append(line)

total_revenue = 0

for member in members:
  total_revenue = total_revenue + member.plan_price()

line = f"Total Revenue: {total_revenue}"
print(line)
report_lines.append(line)

most_visits = members[0]

for member in members:
  if member.visit_count() > most_visits.visit_count():
    most_visits = member

line = f"Member with most visits: {most_visits.name}"
print(line)
report_lines.append(line)

with open("gym_members.txt", "w") as file:
  for line in report_lines:
    file.write(line + "\n")
```
