```python
steps = [6500, 10500, 2500, 9000, 11000, 8000, 12000]
day = 0

while day < len(steps):
  if steps[day] > 10000:
    print("First day crossing 10,000 steps:", day + 1)
    break
  day += 1
```

```python
def ipl_teams():

  teams = ['Mumbai', 'Delhi', 'Chennai', 'Hyderabad', 'Rajasthan', 'Punjab']

  for i in teams:
    # print(i, len(i))

    if len(i) > 6:
      print(i)
      continue

ipl_teams()
```

```python
def song_durations():
    durations = [210, 185, 240, 195, 220]

    for i, duration in enumerate(durations, start=1):
        print(f"Song {i}: {duration} seconds")

song_durations()
```

```python
prices = [250, 65, 0, 800, 2010]

total = 0
for i in prices:
  if i == 0:
    continue

  if total + i > 2000:
    break

  total += i

print("Final Total: ", total)
```
